```python
import numpy as np
from rpy2.robjects import r, FloatVector
import rpy2.robjects.packages as rpackages

# 可选：安装 R 包（只需第一次）
utils = rpackages.importr('utils')
utils.chooseCRANmirror(ind=1)  # 选择镜像
packnames = ('stats',)  # KDE 使用 stats 包
for packname in packnames:
    if not rpackages.isinstalled(packname):
        utils.install_packages(packname)

# 生成示例数据
data = np.random.randn(500)  # 正态分布数据

# 转为 R 的向量
r_data = FloatVector(data)

# 调用 R 的 density 函数计算 KDE
# 这里可以设置 bandwidth, kernel 等参数
density = r['density'](r_data, bw="SJ-dpi", kernel='gaussian')

# 提取 KDE 的 x 和 y
xs = np.array(density.rx2('x'))
ys = np.array(density.rx2('y'))

# 输出前十个点看看
print("X (positions):", xs[:10])
print("Y (density):", ys[:10])
```

```python
# rnaid, chain_id1, chain_id2,...
import os
def generate_config_template(input_dir):
    valid_files = []
    files = os.listdir(input_dir)
    for file in files:
        if file.endswith('.pdb') or file.endswith('.cif'):
            if file not in valid_files:
                valid_files.append(file) # not allowed to put duplicate files, even same id but in different formats
    with open(os.path.join(input_dir, 'config_template.txt'), 'w') as f:
        for file in valid_files:
            rna_ids = file.split('.')[:-1]
            rna_id = ''
            for i in rna_ids:
                rna_id += i
            f.write(f"{rna_id}, \n")
    print("Configuration template generated: config_template.txt, You can fill in chain IDs accordingly.")
            
```

```python
generate_config_template('../data/pdbs')
```
